```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../Data/SampleSuperstore.csv",
    encoding="latin-1"
)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python
# Check missing values in every column
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values.to_frame(name="Missing Count"))
```

```python
# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python
# Store the original row count
rows_before = len(df)

# Remove exact duplicate rows
df = df.drop_duplicates().copy()

# Check the new row count
rows_after = len(df)

print("Rows before cleaning:", rows_before)
print("Duplicate rows removed:", rows_before - rows_after)
print("Rows after cleaning:", rows_after)
print("Remaining duplicate rows:", df.duplicated().sum())
```

```python
df.info
```

```python
print(df.columns.tolist())
```

```python
# Convert sales-related columns to numeric
numeric_columns = [
    "Sales", "Quantity", "Discount", "Profit"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Updated data types:")
print(df[numeric_columns].dtypes)
```

```python
# Remove extra spaces and standardize capitalization
text_columns = [
    "Ship Mode", "Segment", "Country", "City",
    "State", "Region", "Category", "Sub-Category"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip()

print("Text cleaning completed.")
```

```python
print("Dataset shape:", df.shape)
print("Duplicate rows remaining:", df.duplicated().sum())
print("Missing values by column:")

display(df.isnull().sum().to_frame("Missing Count"))
```

```python
# Check numerical data for invalid values

print("Negative Sales:", (df["Sales"] < 0).sum())
print("Zero or negative Quantity:", (df["Quantity"] <= 0).sum())
print("Negative Discount:", (df["Discount"] < 0).sum())
print("Discount greater than 1:", (df["Discount"] > 1).sum())

# Check postal code data type
print("\nPostal Code data type:")
print(df["Postal Code"].dtype)

# Check for any remaining missing values
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Save the cleaned dataset
df.to_csv("../Data/SampleSuperstore_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
