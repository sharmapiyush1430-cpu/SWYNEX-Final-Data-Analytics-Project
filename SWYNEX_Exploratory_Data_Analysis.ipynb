```python
import pandas as pd 
df = pd.read_csv("../Data/SampleSuperstore_cleaned.csv")
df.head()
```

```python
print("Dataset Shape:",df.shape)
```

```python
print("/nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal duplicate rows:")
print(df.duplicated().sum())
```

```python

import matplotlib.pyplot as plt

sales_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(sales_category)

sales_category.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

profit_category = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

print(profit_category)

profit_category.plot(kind="bar", figsize=(8, 5))

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

sales_segment = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

print(sales_segment)

sales_segment.plot(kind="bar", figsize=(8, 5))

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

profit_segment = df.groupby("Segment")["Profit"].sum().sort_values(ascending=False)

print(profit_segment)

profit_segment.plot(kind="bar", figsize=(8, 5))

plt.title("Total Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

subcategory_sales.plot(kind="bar", figsize=(12, 6))

plt.title("Total Sales by Product Sub-Category")
plt.xlabel("Product Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()

subcategory_profit.plot(kind="bar", figsize=(12, 6))

plt.title("Total Profit by Product Sub-Category")
plt.xlabel("Product Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.axhline(y=0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()

print(subcategory_profit)
```

```python

category_discount = df.groupby("Category")["Discount"].mean().sort_values(ascending=False)

print(category_discount)

category_discount.plot(kind="bar", figsize=(8, 5))

plt.title("Average Discount by Product Category")
plt.xlabel("Category")
plt.ylabel("Average Discount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

top_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False).head(10)

print(top_profit)

top_profit.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Product Sub-Categories by Profit")
plt.xlabel("Product Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python

subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values()

print("Product sub-categories ranked by profit:")
print(subcategory_profit)

print("\nLoss-making sub-categories:")
print(subcategory_profit[subcategory_profit < 0])
```

```python

print("Top 10 transactions with the highest discounts:")

print(
    df.nlargest(10, "Discount")[
        ["Sub-Category", "Sales", "Discount", "Profit"]
    ]
)
```

```python

region_performance = df.groupby("Region")[["Sales", "Profit"]].sum()

print(region_performance)

region_performance.plot(
    kind="bar",
    figsize=(10, 5),
    subplots=True,
    layout=(1, 2),
    legend=False
)

plt.suptitle("Sales and Profit by Region")
plt.tight_layout()
plt.show()
```

```python

region_performance["Profit Margin (%)"] = (
    region_performance["Profit"] /
    region_performance["Sales"]
) * 100

print(region_performance.round(2))

region_performance["Profit Margin (%)"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Profit Margin by Region",
    ylabel="Profit Margin (%)",
    xlabel="Region"
)

plt.tight_layout()
plt.show()
```

```python

ship_performance = df.groupby("Ship Mode")[["Sales", "Profit"]].sum()

print("Sales and Profit by Ship Mode:")
print(ship_performance.round(2))

ship_performance.plot(
    kind="bar",
    figsize=(9, 5),
    title="Sales and Profit by Ship Mode"
)

plt.ylabel("Amount")
plt.xlabel("Ship Mode")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```


# Business Insights and Findings

## 1. Regional Performance
The West region has the highest total sales and profit.
The Central region has the lowest profit margin and needs
further investigation.

## 2. Category Performance
Technology has the highest sales among the three product
categories shown in the analysis.

## 3. Loss-Making Products
Tables, Bookcases, and Supplies show negative total profit.
Their pricing, discount policies, and costs should be reviewed.

## 4. Impact of Discounts
The displayed transactions with 80% discounts have negative
profits. High discounts may contribute to losses.

## 5. Customer Segment Performance
The Consumer segment has the highest total sales and profit
among the customer segments shown in the analysis.

## Recommendations
- Investigate the low profit margin in the Central region.
- Review pricing and discounts for loss-making products.
- Maintain strong performance in profitable regions.
- Monitor sales and profit by customer segment.
